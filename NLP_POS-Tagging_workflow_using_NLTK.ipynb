# Program 1: POS Tagging using NLTK

*Objective: POS-tag a set of sentences using NLTK and display the meaning of each tag.*

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

sentences = [
    "The cat is sitting on the mat.",
    "John plays football every evening.",
    "She is reading a book."
]

tag_meanings = {
    "NN": "Noun, singular",
    "NNS": "Noun, plural",
    "NNP": "Proper noun, singular",
    "NNPS": "Proper noun, plural",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund/present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd person singular present",
    "VBZ": "Verb, 3rd person singular present",
    "JJ": "Adjective",
    "JJR": "Adjective, comparative",
    "JJS": "Adjective, superlative",
    "RB": "Adverb",
    "RBR": "Adverb, comparative",
    "RBS": "Adverb, superlative",
    "PRP": "Personal pronoun",
    "DT": "Determiner",
    "IN": "Preposition/subordinating conjunction",
    "CC": "Coordinating conjunction",
    "CD": "Cardinal number",
    "MD": "Modal",
    "TO": "To",
    "RP": "Particle",
    ".": "Punctuation"
}

for sentence in sentences:
    print("\nSentence:", sentence)

    words = nltk.word_tokenize(sentence)
    tagged_words = nltk.pos_tag(words)

    for word, tag in tagged_words:
        meaning = tag_meanings.get(tag, "Other/Unknown tag")
        print(word, "->", tag, "->", meaning)
```

# **Program 2: POS Tagging using spaCy and Comparison with NLTK**

*Objective: POS-tag the same sentences using spaCy and compare the results with NLTK.*

```python
import nltk
import spacy

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

nlp = spacy.load("en_core_web_sm")

sentences = [
    "The cat is sitting on the mat.",
    "John plays football every evening.",
    "She is reading a book."
]

for sentence in sentences:

    print("\nSentence:", sentence)

    words = nltk.word_tokenize(sentence)
    nltk_tags = nltk.pos_tag(words)

    doc = nlp(sentence)

    print("\nNLTK:")
    for word, tag in nltk_tags:
        print(word, "->", tag)

    print("\nspaCy:")
    for token in doc:
        print(token.text, "->", token.pos_, "(", token.tag_, ")")

    print("\nComparison:")
    print("-" * 40)

    for word, nltk_tag in nltk_tags:
        for token in doc:
            if word == token.text:
                print(
                    word,
                    "-> NLTK:", nltk_tag,
                    "| spaCy:", token.pos_, token.tag_
                )
                break
```

# **Program 3: Noun Phrase Chunking using NLTK RegexpParser**

*Objective: Define a chunk grammar and extract noun phrases from a paragraph.*

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

paragraph = """
The young boy is playing football in the beautiful park.
The red car is parked near the big house.
A group of students is studying computer science.
"""

words = nltk.word_tokenize(paragraph)

tagged_words = nltk.pos_tag(words)

print("POS Tagged Words:")
print(tagged_words)

grammar = r"""
    NP: {<DT>?<JJ.*>*<NN.*>+}
"""

chunk_parser = nltk.RegexpParser(grammar)

tree = chunk_parser.parse(tagged_words)

print("\nChunked Tree:")
print(tree)

print("\nNoun Phrases:")

for subtree in tree.subtrees():
    if subtree.label() == "NP":
        phrase = " ".join(word for word, tag in subtree.leaves())
        print(phrase)
```

```python
import nltk
text = ("The old man sat on a "
"wooden bench. A small dog "
"ran across the quiet park.")

sentences = nltk.sent_tokenize(text)
grammar = "NP: {<DT>?<JJ>*<NN>}"
parser = nltk. RegexpParser (grammar)
for sent in sentences:
  tagged = nltk.pos_tag(
      nltk.word_tokenize(sent))
  tree = parser.parse(tagged)
  for sub in tree.subtrees(
      filter=lambda t: t.label()=='NP'):
      print(sub)
```
