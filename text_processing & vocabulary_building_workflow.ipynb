```python
  import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    "machine learning is useful",
    "machine learning is powerful",
    "deep learning is useful",
    "machine learning uses data",
    "data science is useful"
]

document_names = ["D1", "D2", "D3", "D4", "D5"]

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)

vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vocabulary,
    index=document_names
)

print("\nBag-of-Words Matrix:")
print(bow_df)
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_terms = tfidf_vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_terms,
    index=document_names
)

print("TF-IDF Matrix:")
print(tfidf_df.round(4))

print("\nTop-weighted term for each document:")

for i, document in enumerate(document_names):
    row = tfidf_matrix[i].toarray().flatten()
    top_index = row.argmax()

    print(
        f"{document} → Term: {tfidf_terms[top_index]} "
        f"→ TF-IDF: {row[top_index]:.4f}"
    )
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)

similarity_df = pd.DataFrame(
    similarity_matrix,
    columns=document_names,
    index=document_names
)

print("Cosine Similarity Matrix:")
print(similarity_df.round(4))

pairs = []

for i in range(len(document_names)):
    for j in range(i + 1, len(document_names)):
        pairs.append(
            (
                document_names[i],
                document_names[j],
                similarity_matrix[i][j]
            )
        )

pairs_sorted = sorted(pairs, key=lambda x: x[2], reverse=True)

most_similar = pairs_sorted[0]
least_similar = pairs_sorted[-1]

print("\nMost Similar:")
print(f"{most_similar[0]} and {most_similar[1]} → {most_similar[2]:.4f}")

print("\nLeast Similar:")
print(f"{least_similar[0]} and {least_similar[1]} → {least_similar[2]:.4f}")
```

```python
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(documents)

bow_terms = count_vectorizer.get_feature_names_out()

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_terms,
    index=document_names
)

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_terms = tfidf_vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_terms,
    index=document_names
)

print("BAG-OF-WORDS MATRIX")
print(bow_df)

print("\nTF-IDF MATRIX")
print(tfidf_df.round(4))
```

```python
common_words = ["machine", "learning", "is", "useful", "data"]

print("Comparison of BoW and TF-IDF values:\n")

for word in common_words:
    print(f"Word: {word}")
    print("BoW:")
    print(bow_df[word])
    print("\nTF-IDF:")
    print(tfidf_df[word].round(4))
    print("-" * 40)
```

```python
print("Top Two TF-IDF Terms for Each Document:\n")

for i, document in enumerate(document_names):

    row = tfidf_matrix[i].toarray().flatten()

    top_two_indices = row.argsort()[-2:][::-1]

    term1 = tfidf_terms[top_two_indices[0]]
    score1 = row[top_two_indices[0]]

    term2 = tfidf_terms[top_two_indices[1]]
    score2 = row[top_two_indices[1]]

    print(
        f"{document} → "
        f"{term1} ({score1:.4f}), "
        f"{term2} ({score2:.4f})"
    )
```

```python
d1_index = 0

print("Similarity of D1 with other documents:\n")

d1_similarities = []

for i in range(1, len(document_names)):

    score = similarity_matrix[d1_index][i]

    d1_similarities.append(
        (document_names[i], score)
    )

    print(
        f"D1 vs {document_names[i]} → {score:.4f}"
    )

d1_similarities.sort(key=lambda x: x[1], reverse=True)

print("\nRanking:")
for rank, (doc, score) in enumerate(d1_similarities, start=1):
    print(f"{rank}. {doc} → {score:.4f}")

print("\nMost Similar →", d1_similarities[0][0])
print("Least Similar →", d1_similarities[-1][0])
```

```python
documents_q7 = [
    "machine learning algorithms analyze data",
    "deep learning models process large datasets",
    "football players train for important matches",
    "data science uses machine learning techniques",
    "healthy food and regular exercise improve health",
    "artificial intelligence and machine learning use data"
]

document_names_q7 = [
    "Document 1",
    "Document 2",
    "Document 3",
    "Document 4",
    "Document 5",
    "Document 6"
]

query = "machine learning data"

vectorizer_q7 = TfidfVectorizer()

document_vectors = vectorizer_q7.fit_transform(documents_q7)

query_vector = vectorizer_q7.transform([query])

similarities = cosine_similarity(
    query_vector,
    document_vectors
).flatten()

ranking = similarities.argsort()[::-1]

print("Query:", query)
print("\nDocument Ranking:\n")

for rank, index in enumerate(ranking, start=1):
    print(
        f"Rank {rank} → {document_names_q7[index]} "
        f"→ Similarity: {similarities[index]:.4f}"
    )
```

```python
news_documents = [
    "football team won the match after excellent performance by players",
    "new technology company launched an artificial intelligence smartphone",
    "students attended online classes and completed their education courses",
    "doctors discussed healthy diet exercise and disease prevention",
    "technology companies invested in artificial intelligence and software development"
]

news_names = [
    "Sports",
    "Technology",
    "Education",
    "Health",
    "Business"
]

news_vectorizer = TfidfVectorizer()

news_tfidf = news_vectorizer.fit_transform(news_documents)

news_terms = news_vectorizer.get_feature_names_out()

news_tfidf_df = pd.DataFrame(
    news_tfidf.toarray(),
    columns=news_terms,
    index=news_names
)

print("TF-IDF Matrix:")
print(news_tfidf_df.round(4))

news_similarity = cosine_similarity(news_tfidf)

news_similarity_df = pd.DataFrame(
    news_similarity,
    columns=news_names,
    index=news_names
)

print("\nPairwise Cosine Similarity:")
print(news_similarity_df.round(4))
```

```python
news_pairs = []

for i in range(len(news_names)):
    for j in range(i + 1, len(news_names)):
        news_pairs.append(
            (
                news_names[i],
                news_names[j],
                news_similarity[i][j]
            )
        )

news_pairs.sort(key=lambda x: x[2], reverse=True)

most_similar_news = news_pairs[0]

print("\nMost Similar Article Pair:")
print(
    f"{most_similar_news[0]} and {most_similar_news[1]}"
    f" → Similarity: {most_similar_news[2]:.4f}"
)
```

```python
documents_q9 = [
    "machine machine machine learning learning data",
    "machine learning data science",
    "machine learning data analysis",
    "machine learning data research",
    "rareword technology research"
]

names_q9 = ["D1", "D2", "D3", "D4", "D5"]

vectorizer_q9 = TfidfVectorizer()

tfidf_q9 = vectorizer_q9.fit_transform(documents_q9)

terms_q9 = vectorizer_q9.get_feature_names_out()

tfidf_q9_df = pd.DataFrame(
    tfidf_q9.toarray(),
    columns=terms_q9,
    index=names_q9
)

print("TF-IDF Matrix:")
print(tfidf_q9_df.round(4))
```

```python
selected_words = ["machine", "learning", "rareword", "technology"]

print("\nSelected TF-IDF Values:\n")

for word in selected_words:
    if word in tfidf_q9_df.columns:
        print(f"{word}:")
        print(tfidf_q9_df[word].round(4))
        print()
```

```python
print("Maximum TF-IDF score of selected words:\n")

for word in selected_words:
    if word in tfidf_q9_df.columns:
        max_score = tfidf_q9_df[word].max()
        document = tfidf_q9_df[word].idxmax()

        print(
            f"{word} → {max_score:.4f} in {document}"
        )
```
